```python
%sql
CREATE CATALOG IF NOT EXISTS ecommerce_lakehouse;
```

```python
%sql
USE CATALOG ecommerce_lakehouse;

CREATE SCHEMA IF NOT EXISTS bronze
COMMENT 'Raw ingestion layer containing landing delta tables with original source schemas and metadata.';

CREATE SCHEMA IF NOT EXISTS silver
COMMENT 'Cleaned, conformed, and deduplicated enterprise data layer.';

CREATE SCHEMA IF NOT EXISTS gold
COMMENT 'Aggregated dimensional modeling layer (Star Schema) optimized for BI and analytics.';
```

```python
%sql
CREATE VOLUME IF NOT EXISTS bronze.raw_files
COMMENT 'Volume for landing raw CSV files before ingestion into Bronze Delta tables.';
```
